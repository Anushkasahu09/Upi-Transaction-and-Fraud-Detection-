```python
# Create table (example)
import mysql.connector

# MySQL se connect
connection = mysql.connector.connect(
    host="localhost",
    user="root",
    password="0912"
)

cursor = connection.cursor()

# Database create karo
cursor.execute("CREATE DATABASE IF NOT EXISTS upi_fraud")

# Database select karo
cursor.execute("USE upi_fraud")

# Check karo kaunsa database select hua
cursor.execute("SELECT DATABASE()")
print(cursor.fetchone())

# Table create karo
cursor.execute("""
CREATE TABLE IF NOT EXISTS transactions (
    txn_id VARCHAR(50),
    user_id VARCHAR(50),
    amount DECIMAL(10,2)
)
""")

connection.commit()
print("Table created successfully!")
```

```python
#Create the transactions table
cursor.execute("""
CREATE TABLE IF NOT EXISTS transactions (
    transaction_id VARCHAR(50),
    timestamp DATETIME,
    transaction_type VARCHAR(50),
    merchant_category VARCHAR(100),
    amount DECIMAL(10,2),
    transaction_status VARCHAR(50),
    sender_age_group VARCHAR(20),
    receiver_age_group VARCHAR(20),
    sender_state VARCHAR(50),
    sender_bank VARCHAR(100),
    receiver_bank VARCHAR(100),
    device_type VARCHAR(50),
    network_type VARCHAR(50),
    fraud_flag INT,
    hour_of_day INT,
    day_of_week VARCHAR(20),
    is_weekend BOOLEAN
)
""")

connection.commit()
print("Transactions table created successfully!")
```

```python
query = """
INSERT INTO transactions (
    transaction_id,
    timestamp,
    transaction_type,
    merchant_category,
    amount,
    transaction_status,
    sender_age_group,
    receiver_age_group,
    sender_state,
    sender_bank,
    receiver_bank,
    device_type,
    network_type,
    fraud_flag,
    hour_of_day,
    day_of_week,
    is_weekend
)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
"""

for _, row in df.iterrows():
    cursor.execute(query, 
        (row["transaction id"],       # CSV column has a space
        row["timestamp"].to_pydatetime() if pd.notna(row["timestamp"]) else None,
        row["transaction type"],      # CSV column has a space
        row["merchant_category"],
        float(row["amount (INR)"]),
        row["transaction_status"],
        row["sender_age_group"],
        row["receiver_age_group"],
        row["sender_state"],
        row["sender_bank"],
        row["receiver_bank"],
        row["device_type"],
        row["network_type"],
        int(row["fraud_flag"]),
        int(row["hour_of_day"]),
        row["day_of_week"],
        bool(row["is_weekend"])
    ))

connection.commit()

print("✅ All data inserted successfully!")
```
